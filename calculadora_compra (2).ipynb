```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Título del formulario adaptado a la actividad
titulo = widgets.HTML("<h2>CALCULADORA DE COMPRA</h2>")

# 2. Campos de entrada de datos
txt_producto = widgets.Text(
    description="Producto:",
    placeholder="Escriba el nombre del producto"
)

txt_precio = widgets.FloatText(
    description="Precio (Bs):",
    value=0.0
)

txt_cantidad = widgets.IntText(
    description="Cantidad:",
    value=1
)

# Botón para calcular el total
btn_calcular = widgets.Button(
    description="Calcular Compra",
    button_style="success",
    icon="calculator"
)

# Área donde se mostrará el resumen
salida = widgets.Output()

# 3. Función para procesar los datos y calcular el costo total
def calcular_compra(b):
    with salida:
        clear_output(wait=True)

        # Captura de datos desde la interfaz gráfica
        producto = txt_producto.value.strip()
        precio_unitario = txt_precio.value
        cantidad = txt_cantidad.value

        # Validaciones de seguridad
        if not producto:
            print("⚠️ Error: Por favor, ingrese el nombre del producto.")
            return

        if precio_unitario <= 0:
            print("⚠️ Error: El precio unitario debe ser mayor a 0.")
            return

        if cantidad <= 0:
            print("⚠️ Error: La cantidad debe ser al menos 1.")
            return

        # Cálculo del costo total
        costo_total = precio_unitario * cantidad

        # 4. Impresión del resumen de la compra
        print("\n" + "="*35)
        print("        RESUMEN DE COMPRA        ")
        print("="*35)
        print(f"Producto:         {producto}")
        print(f"Precio Unitario:  ${precio_unitario:,.2f}")
        print(f"Cantidad:         {cantidad}")
        print("-"*35)
        print(f"COSTO TOTAL:      ${costo_total:,.2f}")
        print("="*35)

# Asociar el evento de clic del botón con la función de cálculo
btn_calcular.on_click(calcular_compra)

# 5. Mostrar los elementos de la interfaz en Google Colab
display(
    titulo,
    txt_producto,
    txt_precio,
    txt_cantidad,
    btn_calcular,
    salida
)
```
